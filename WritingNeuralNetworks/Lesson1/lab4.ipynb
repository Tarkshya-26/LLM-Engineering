## LESSON 4 - How Does Bob Learn From His Mistakes?

SO far We've Covered that a Machine can :

## 1. Make predictions.
## 2. Compare predictions with the actual answers.
## 3. Calculate how wrong its predictions are using Mean Squared Error (MSE).



## The Problem - He knows that he's wrong but doesn't know how to fix it???


## But here's the catch that we chose those weights ourselves

```python
## Let Bob(Our Neuron) run the experiments 
import numpy as np

hours = np.array([1, 2, 3, 4], dtype=float)
marks = np.array([20, 40, 60, 80], dtype=float)
weights = np.arange(0,31,1)

for weight in weights:
    predictions = hours * weight
    mse = np.mean((marks - predictions) ** 2)

    print(f"Weight: {weight:2d} | MSE: {mse:7.1f}")


```

## HERE WE GET THE ANSWER AS 20 with 0 MSE

```python
weight = 5.0
step = 0.01

predictions_1 = hours * weight
loss_1 = np.mean((marks - predictions_1) ** 2)

predictions_2 = hours * (weight + step)
loss_2 = np.mean((marks - predictions_2) ** 2)

slope = (loss_2 - loss_1) / step

print("Loss1 at weight 5:", loss_1)
print("Loss2 at weight 5.01:", loss_2)
print("Approximate slope:", slope)
```

The approximate slope will be negative.
That tells us that increasing the weight slightly decreases the loss.

## Method 1:[Value low]() (+) Weight  (-)LOSS 

```python
## now trying the same experiment at weight 25 
weight = 25.0
step = 0.01

predictions_1 = hours * weight
loss_1 = np.mean((marks - predictions_1) ** 2)

predictions_2 = hours * (weight + step)
loss_2 = np.mean((marks - predictions_2) ** 2)

slope = (loss_2 - loss_1) / step

print("Loss1 at weight 5:", loss_1)
print("Loss2 at weight 5.01:", loss_2)
print("Approximate slope:", slope)
```

The slope will be positive.
That tells us that increasing the weight makes the loss worse, so we should move in the opposite direction.

## Method 2:[Value High](***)(+) WEIGHT (+) LOSS , Therefore Lower the value of weight . Here + - means increasing and decreasing

## SLOPE INVESTIGATION : THE IDEAL WEIGHT is: [20](+) | for [0](-) MSE
