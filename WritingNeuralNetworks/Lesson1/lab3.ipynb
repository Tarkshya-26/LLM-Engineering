```python
"""Why would anyone square the error?
Excellent question.
Let's compare.
Prediction A
79

Error
1

Squared
1

Prediction B
40

Error
40

Squared
1600

Notice what happened?
Large mistakes became much larger.
Small mistakes stayed small.
This is exactly what we want.
We want Bob to care much more about huge mistakes than tiny ones."""
```

```python
import numpy as np

hours = np.array([1,2,3,4], dtype=float)

marks = np.array([20,40,60,80], dtype=float)

weight = 5.0

predictions = hours * weight

errors = marks - predictions

print(errors)
```

```python
## NOW SQUARE THE ERRORS

squared_errors = errors ** 2

print(squared_errors)
```

## How badly did our neuron performed ?

```python
squared_errors = np.array([225, 900, 2025, 3600])

average = np.mean(squared_errors)

print(average)
```

```python
"""Why don't we simply use the total?
225 + 900 + 2025 + 3600


Good question.
Suppose tomorrow we collect data from
4 students

The total error is
6750

Next week we collect data from
4000 students

Now the total becomes
8,000,000

Did Bob suddenly become worse?
Not necessarily.
He simply had more examples.
Taking the mean makes the loss independent of how many training examples we have.
That's why averaging is preferred."""
```
