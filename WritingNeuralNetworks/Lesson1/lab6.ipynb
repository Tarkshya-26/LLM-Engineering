## Lesson 6 : Training Bob with Graident Descent 



```python
import numpy as np

hours = np.array([1, 2, 3, 4], dtype=float)
marks = np.array([20, 40, 60, 80], dtype=float)

weight = 10.0
eta = 0.05
epochs = 100

for epoch in range(epochs):

    # 1. Forward pass: make predictions
    predictions = hours * weight

    # 2. Calculate loss
    loss = np.mean((predictions - marks) ** 2)

    # 3. Calculate the gradient
    gradient = np.mean(2 * (predictions - marks) * hours)

    # 4. Update the weight 
    weight = weight - eta * gradient

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.2f}, Weight = {weight:.4f}")
```

## Key takeaway: A single gradient update is a learning step. Repeating those steps across epochs is how Bob learns from the data.
