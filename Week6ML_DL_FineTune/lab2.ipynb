```python
username = "ed-donner"
dataset = f"{username}/items_raw_lite" if LITE_MODE else f"{username}/items_raw_full"

train, val, test = Item.from_hub(dataset)

items = train + test + val

print(f"loaded {len(items):,} items")
print(items[0])
```

```python
items[2].id
```

```python
# Give every item an id 

for index, item in enumerate(items):
    item.id = index
```

```python


SYSTEM_PROMPT = """Create a concise description of a product. Respond only in this format. Do not include part numbers.
Title: Rewritten short precise title
Category: eg Electronics
Brand: Brand name
Description: 1 sentence description
Details: 1 sentence on features"""
```

```python
print(items[0].full)
```

```python
messages = [{"role":"system","content":SYSTEM_PROMPT},{"role":"user", "content":items[0].full}]

response = completion(messages= messages, model = "groq/openai/gpt-oss-20b",reasoning_effort="low")
print(response.choice[0].message.content)
print()
print(f"Input tokens: {response.usage.prompt_tokens}")
print(f"Output tokens: {response.usage.completion_tokens}")
print(f"Cost: {response._hidden_params['response_cost']*100:.3f} cents")
```

```python
"""6. Why does he test Ollama?

He then does:

model="ollama/llama3.2"
api_base="http://localhost:11434"

This demonstrates that the transformation doesn't fundamentally depend on one specific provider.

Conceptually:

                SYSTEM PROMPT
                      +
                  ITEM.FULL
                      │
            ┌─────────┴─────────┐
            ▼                   ▼
          Groq                Ollama
            │                   │
            ▼                   ▼
        LLM output          LLM output

You can therefore experiment with:

hosted model
local model
potentially another provider

while keeping the data transformation task the same."""
```

```python
# MAKING CALLS WITH LOCAL MODEL

messages = [{"role":"system","content":SYSTEM_PROMPT},{"role":"user","content":items[0].full}]
response = completion(messages=messages, model = "ollama/llama3.2",api_base = "http://localhost:11434")
print(response.choices[0].message.content)
print()
print(f"Input tokens: {response.usage.prompt_tokens}")
print(f"Output tokens: {response.usage.completion_tokens}")
print(f"Cost: {response._hidden_params['response_cost']*100:.3f} cents")
```

```python
"""5. Why does Ed call the LLM individually first?

He tests:

messages = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": items[0].full}
]

response = completion(...)

This is basically:

SYSTEM PROMPT
      +
PRODUCT DATA
      ↓
     LLM
      ↓
STANDARDIZED PRODUCT DESCRIPTION

He also prints:

response.usage.prompt_tokens
response.usage.completion_tokens
response._hidden_params['response_cost']

This is important because LLM preprocessing has a cost.

Before launching thousands of requests, you want to know:

How many tokens am I sending?
How many tokens am I generating?
How much does one transformation cost?

Otherwise you could accidentally launch:

800,000 × expensive LLM call

without understanding the bill."""
```

```python
MODEL = "openai/gpt-oss-20b"
```

```python
def make_jsonl(item):
    body = {"model":MODEL,"messages":[{"role":"system","content":SYSTEM_PROMPT},{"role":"user","content":item.full}],"reasoning_effort":"low"}
    line = {"custom_id":str(item.id),"method":"POST","url":"/v1/chat/completions","body":body}
    return json.dumps(line)
```

```python
items[0]
```

```python
make_jsonl(items[0])
```

```python
def make_file(start, end, filename):
    batch_file = filename
    with open(batch_file, "w", encoding="utf-8") as f:
        for i in range(start, end):
            f.write(make_jsonl(items[i]))
            f.write("\n")
```

```python
make_file(0, 1000, "jsonl/0_1000.jsonl")
```

```python
import os
from groq import Groq

groq = Groq(api_key=os.environ.get("GROQ_API_KEY"))
```

```python

with open("jsonl/0_1000.jsonl", "rb", encoding="utf-8") as f:
    response = groq.files.create(file=f, purpose="batch")
response
```

```python
file_id = response.id
file_id
```

```python
response = groq.batches.create(completion_window="24h", endpoint="/v1/chat/completions", input_file_id=file_id)
response
```

```python
result = groq.batches.retrieve(response.id)
result
```

```python
response = groq.files.content(result.output_file_id)
response.write_to_file("jsonl/batch_results.jsonl")
```
