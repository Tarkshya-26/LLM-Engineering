## What is np.array()?
Definition
np.array() creates a NumPy Array, a data structure optimized for numerical and mathematical computations.

## Common Mistake:
Thinking NumPy eliminates loops. It doesn't—the loops still exist, but they run inside optimized compiled code rather than as Python for loops.
## Key Takeaway:
Vectorization lets you express mathematical operations on whole arrays, while NumPy executes them efficiently under the hood using optimized low-level code.
