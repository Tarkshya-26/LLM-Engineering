# The Executive Summary

# **The KEY POINTS TO REMEMBER**
STORAGE DESIGN FOR PRODUCTION RAG
─────────────────────────────────
<pre>
1. Vector DB ≠ source of truth
   → It is a retrieval projection/shadow.

2. Metadata
   → Keep fields that support:
      filtering
      governance
      refresh
      debugging
      evaluation

3. Filters
   → Shrink candidate set BEFORE ranking.
   → Bad metadata can destroy recall.

4. Collections
   → One = simpler
   → Many = stronger domain isolation
   → Decide based on operational/security boundaries.

5. Content storage
   → Snippet = faster, but duplication/staleness risk.
   → Reference = fresher, but potentially slower.
   → Hybrid = chunk preview + authoritative reference.

6. Checksum
   → Content fingerprint for change detection.

7. Production schema
   → identity + lineage + version + governance
      + operational metadata.

8. Design principle
   → Every field should support an actual decision.
   </pre>
